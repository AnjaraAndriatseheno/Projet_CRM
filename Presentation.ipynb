PROBLEMATQUE

Un cinquième de nos clients porte les deux tiers de notre chiffre d'affaires
— et on ne sait ni pourquoi ils nous quittent, ni si nos campagnes
marketing les touchent vraiment.

Comment sécuriser durablement notre CA en agissant à la fois sur
la rétention client et sur nos investissements marketing ?

On a analysé l'ensemble de notre base client — 49 402 comptes — croisée avec plus d'un million de points de contact marketing sur nos 6 dernières campagnes. Un constat frappant en ressort immédiatement : un cinquième de nos clients porte les deux tiers de notre chiffre d'affaires. Et on ne sait aujourd'hui ni pourquoi une partie d'entre eux nous quitte, ni si nos campagnes les touchent efficacement. C'est la question qu'on va résoudre aujourd'hui."

HYPOTHESES

- Les clients à risque partagent un un achat commun avant décrochage
- Les clients internationaux décrochent davantage que les clients français, peut-être en raison d'une expérience moins adaptée à leur marché (livraison, communication
- Les clients À risque ont reçu moins de points de contact marketing dans la période précédant leur inactivité, comparé aux clients toujours actifs — ils décrochent peut-être parce qu'on arrête de les solliciter, pas l'inverse.

RECOMMANDATATIONS

1. Les clients "à risque" partagent un achat commun avant décrochage — confirmé Le dernier achat des clients à risque contient seulement ~10% de la catégorie "Coffret", contre ~21% dans le catalogue global — c'est l'écart le plus marqué de toutes les catégories (les autres restent proches de leur poids naturel). Le signal est réel, pas un artefact de popularité générale.

- Action : cibler en cross-sell les clients qui viennent d'acheter mono-catégorie (Corps/Cheveux/Bain/Accessoire) sans "Coffret" avec une recommandation automatique de coffret découverte — c'est exactement le pattern absent chez ceux qui décrochent.

- Action : traiter le Coffret comme un produit de rétention à pousser dans les campagnes de fidélisation, pas seulement comme un article cadeau saisonnier.

- Avant de généraliser le budget : un test A/B (proposer un coffret à un échantillon de clients récents mono-catégorie vs un groupe témoin) permettrait de confirmer que c'est bien causal et pas juste corrélé à un profil de client déjà plus engagé.



2. Les clients internationaux décrochent davantage — non confirmé, l'analyse déjà faite dit l'inverse Le graphique déjà produit en TP2 (comparaison À risque/Perdus/Champions/Fidèles par géographie) montre que les clients à risque et perdus sont légèrement plus fréquents côté France (7,7% et 6,4%) que côté international (5,7% et 5,6%) — l'exact opposé de l'hypothèse.

- Action : ne pas lancer d'investissement (livraison, communication localisée) sur la base de cette hypothèse telle quelle — les données déjà calculées ne la soutiennent pas.

- Action : avant toute conclusion, vérifier la robustesse statistique de cet écart (la base internationale ne pèse que ~8% du total, donc l'intervalle de confiance est large — l'écart pourrait ne pas être significatif dans un sens ou dans l'autre).

- Si tu veux vraiment creuser le sujet livraison/communication, il faudrait un indicateur direct (délai de livraison moyen par pays, langue des campagnes reçues) est - ce que les données actuelles ne permettent pas d'évaluer ; le crosstab segment × pays ne mesure que le résultat, pas la cause.



3. Moins de points de contact marketing avant le décrochage — confirmé, et c'est le signal le plus fort des troisDans les 90 jours précédant leur dernière activité, les clients à risque reçoivent en moyenne 0,93 touchpoint, contre 14,1 pour les Champions et 9,7 pour les Fidèles (soit 15 fois moins) ; les Perdus, 0.

- Action : mettre en place une alerte CRM automatique — dès qu'un client passe sous un seuil de sollicitations (ex : moins de 3 touchpoints sur 60 jours) sans être encore "Perdu", déclencher une relance avant qu'il ne bascule.

- Action prioritaire : tester la causalité avant de scaler, via un A/B test — maintenir un niveau de sollicitation proche de celui des Fidèles sur un échantillon de clients en début de désengagement, et comparer leur taux de bascule vers "À risque" à un groupe témoin laissé au traitement actuel. Ça tranchera si "moins sollicité → décroche" ou si c'est l'inverse (le ciblage délaisse déjà ceux qui montrent des signes de désintérêt).

Vérifier aussi que le système d'exclusion marketing (opt-out, listes de suppression) n'exclut pas par erreur des clients récupérables — vu que "Perdus" = 0 touchpoint pile, ça vaut la peine de s'assurer que ce n'est pas un artefact technique.
